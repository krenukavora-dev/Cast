```
import pandas as pd
cast_df = pd.read_csv('/content/drive/MyDrive/cast.csv')
display(cast_df.head())
```

```
# How many movies are listed in this dataframe?
print (len(cast_df))
```

```
print (cast_df.shape[0])
```

```
# What are the earliest two films listed in the titles dataframe?
print (cast_df.head(2))
```

```
# How many movies have the title "Hamlet"?
print ((cast_df['title']=="Hamlet").sum())
```

```
# How many movies are titled "North by Northwest"?
print ((cast_df['title']=="North by Northwest").sum())
```

```
# When was the first movie titled "Hamlet" made?
hamlet_movies = cast_df[cast_df['title'] == 'Hamlet']
first_hamlet_year = hamlet_movies['year'].min()
print(f"The first movie titled 'Hamlet' was made in {int(first_hamlet_year)}.")
```

```
# List all of the 'Treasure Island' movies from earliest to the most recent?
treasure_island_movies = cast_df[cast_df['title'] == 'Treasure Island'].sort_values('year').drop_duplicates('year')
for index, row in treasure_island_movies.iterrows():
    print(f"{int(row['year'])}: {row['title']}")
```

```
# How many movies were made in the year 1950?
movies_1950 = cast_df[cast_df['year'] == 1950]['title'].nunique()
print(f"There were {int(movies_1950)} unique movies made in the year 1950.")
```

```
# How many movies were made in the year 1960?
movies_1960 = cast_df[cast_df['year'] == 1960]['title'].nunique()
print(f"There were {int(movies_1960)} unique movies made in the year 1960.")
```

```
# How many movies were made from 1950 through 1959?
movies_1950_1959 = cast_df[(cast_df['year'] >= 1950) & (cast_df['year'] <= 1959)]['title'].nunique()
print(f"There were {int(movies_1950_1959)} unique movies made from 1950 through 1959.")
```

```
# In what years has a movie titled "Batman" been released?
batman_movies_years = cast_df[cast_df['title'] == 'Batman']['year'].unique()
print(f"Movies titled 'Batman' have been released in the years: {sorted(batman_movies_years)}.")
```

```
# How many roles were there in the movie "Inception"?
inception_total_roles = cast_df[cast_df['title'] == 'Inception'].shape[0]
print(f"There are {inception_total_roles} total roles in the movie 'Inception'.")
```

```
# How many roles in the movie "Inception" are NOT ranked by an "n" value?
inception_roles = cast_df[cast_df['title'] == 'Inception']
unranked_inception_roles = inception_roles['n'].isnull().sum()
print(f"There are {int(unranked_inception_roles)} roles in 'Inception' that are not ranked by an 'n' value.")
```

```
#But how many roles in the movie "Inception" did receive an "n" value?
inception_roles = cast_df[cast_df['title'] == 'Inception']
ranked_inception_roles = inception_roles['n'].notnull().sum()
print(f"There are {int(ranked_inception_roles)} roles in 'Inception' that received an 'n' value.")
```

```
# Display the cast of "North by Northwest" in their correct "n"- value order, ignoring roles that did not earn a numeric "n" value?
north_by_northwest_cast = cast_df[(cast_df['title'] == 'North by Northwest') & (cast_df['n'].notnull())].sort_values('n')
print("Cast of 'North by Northwest' (ordered by 'n' value):")
for index, row in north_by_northwest_cast.iterrows():
    print(f"  {row['name']} as {row['character']} (n={int(row['n'])})")
```

```
#Display the entire cast, in "n"- order, of the 1972 film "Sleuth"?
Sleuth_1972 = cast_df[(cast_df['title'] == 'Sleuth') & (cast_df['year'] == 1972) & (cast_df['n'].notnull())].sort_values('n')
print("Cast of 'Sleuth' (1972, ordered by 'n' value):")
for index, row in Sleuth_1972.iterrows():
    print(f"  {row['name']} as {row['character']} (n={int(row['n'])})")
```

```
# Now display the entire cast, in "n"- order, of the 2007 version of "Sleuth"?
Sleuth_2007 = cast_df[(cast_df['title'] == 'Sleuth') & (cast_df['year'] == 2007) & (cast_df['n'].notnull())].sort_values('n')
print("Cast of 'Sleuth' (2007, ordered by 'n' value):")
for index, row in Sleuth_2007.iterrows():
    print(f"  {row['name']} as {row['character']} (n={int(row['n'])})")
```

```
# How many roles were credited in the silent 1920 version of Hamlet?
hamlet_movies = cast_df[cast_df['title'] == 'Hamlet']
silent_1920_hamlet_roles_df = hamlet_movies[hamlet_movies['year'] == 1920]
num_silent_hamlet_roles = silent_1920_hamlet_roles_df.shape[0]
print(f"There were {num_silent_hamlet_roles} roles in the 1920 version of 'Hamlet'.")
```

```
# How many roles were credited in Branagh's 1996 Hamlet?
branagh_hamlet_roles = cast_df[(cast_df['title'] == 'Hamlet') & (cast_df['year'] == 1996)]
num_branagh_hamlet_roles = branagh_hamlet_roles.shape[0]
print(f"There were {num_branagh_hamlet_roles} roles credited in Branagh's 1996 'Hamlet'.")
```

```
# How many "Hamlet" roles have been listed in all film credits through history?
hamlet_roles_count = cast_df[cast_df['title'] == 'Hamlet'].shape[0]
print(f"There have been {hamlet_roles_count} roles listed in all 'Hamlet' film credits through history.")
```

```
# How many people have played an "Ophelia"?
ophelia_roles = cast_df[cast_df['character'] == 'Ophelia']
unique_ophelias = ophelia_roles['name'].nunique()
print(f"There are {unique_ophelias} people who have played the role of 'Ophelia'.")
```

```
# How many people have played a role called "The Dude"?
dude_roles = cast_df[cast_df['character'] == 'The Dude']
unique_dudes = dude_roles['name'].nunique()
print(f"There are {unique_dudes} people who have played the role of 'The Dude'.")
```

```
# How many people have played the role called "The Stranger"?
stranger_roles = cast_df[cast_df['character'] == 'The Stranger']
unique_strangers = stranger_roles['name'].nunique()
print(f"There are {unique_strangers} people who have played the role of 'The Stranger'.")
```

```
# How many role has Sidney Poitier played throughtout his career?
sidney_roles = cast_df[cast_df['name'] == 'Sidney Poitier']
num_sidney_roles = sidney_roles.shape[0]
print(f"Sidney Poitier has played {num_sidney_roles} roles throughout his career.")
```

```
# How many roles has Judi Dench played?
judi_dench_roles = cast_df[cast_df['name'] == 'Judi Dench']
num_judi_dench_roles = judi_dench_roles.shape[0]
print(f"Judi Dench has played {num_judi_dench_roles} roles throughout her career.")
```

```
# List the supporting roles (having n=2) played by Cary Grant in the 1940s, in order by year?
cary_grant_supporting_1940s = cast_df[
    (cast_df['name'] == 'Cary Grant') &
    (cast_df['n'] == 2) &
    (cast_df['year'] >= 1940) &
    (cast_df['year'] <= 1949)
].sort_values('year')

print("Cary Grant's supporting roles (n=2) in the 1940s:")
if not cary_grant_supporting_1940s.empty:
    for index, row in cary_grant_supporting_1940s.iterrows():
        print(f"  {int(row['year'])}: {row['title']} as {row['character']}")
else:
    print("  No supporting roles (n=2) found for Cary Grant in the 1940s.")
```

```
# List the leading roles that Cary Grant played in the 1940s in order by year?
cary_grant_leading_1940s = cast_df[
    (cast_df['name'] == 'Cary Grant') &
    (cast_df['n'] == 1) &
    (cast_df['year'] >= 1940) &
    (cast_df['year'] <= 1949)
].sort_values('year')

print("Cary Grant's leading roles (n=1) in the 1940s:")
if not cary_grant_leading_1940s.empty:
    for index, row in cary_grant_leading_1940s.iterrows():
        print(f"  {int(row['year'])}: {row['title']} as {row['character']}")
else:
    print("  No leading roles (n=1) found for Cary Grant in the 1940s.")
```

```
# How many roles were available for actors in the 1950s?
actors_1950s = cast_df[(cast_df['type'] == 'actor') & (cast_df['year'] >= 1950) & (cast_df['year'] <= 1959)]
num_actors_roles_1950s = actors_1950s.shape[0]
print(f"There were {num_actors_roles_1950s} roles available for actors in the 1950s.")
```

```
# How many roles were available for actresses in the 1950s?
actresses_1950s = cast_df[(cast_df['type'] == 'actress') & (cast_df['year'] >= 1950) & (cast_df['year'] <= 1959)]
num_actresses_roles_1950s = actresses_1950s.shape[0]
print(f"There were {num_actresses_roles_1950s} roles available for actresses in the 1950s.")
```

```
# How many leading roles (n=1) were available from the beginning of film history through 1980?
leading_roles_through_1980 = cast_df[(cast_df['n'] == 1) & (cast_df['year'] <= 1980)]
num_leading_roles_through_1980 = leading_roles_through_1980.shape[0]
print(f"There were {num_leading_roles_through_1980} leading roles (n=1) available from the beginning of film history through 1980.")
```

```
# How many non-leading roles were available from the beginning of film history through 1980?
non_leading_roles_through_1980 = cast_df[(cast_df['n'] > 1) & (cast_df['year'] <= 1980)]
num_non_leading_roles_through_1980 = non_leading_roles_through_1980.shape[0]
print(f"There were {num_non_leading_roles_through_1980} non-leading roles (n > 1) available from the beginning of film history through 1980.")
```

```
# How many roles through 1980 were minor enough that they did not warrant a numeric "n" rank?
unranked_roles_through_1980 = cast_df[(cast_df['n'].isnull()) & (cast_df['year'] <= 1980)]
num_unranked_roles_through_1980 = unranked_roles_through_1980.shape[0]
print(f"There were {num_unranked_roles_through_1980} roles through 1980 that did not warrant a numeric 'n' rank.")
```
